```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import json
import validation
```

# Calificación de respuestas del sistema

Este notebook realiza un análisis de las respuestas del sistema para evaluar su precisión, utilizando una escala de valoración del 2 al 5. Además, se pude examinar si la dificultad de las preguntas influye en la exactitud de las respuestas.


```python
preguntas = validation.cargar_json("questions.json")
validation.mostrar_preguntas(preguntas)
```

```python
data = validation.cargar_json("result.json")  # Reemplaza `json_string` con tu JSON
df = pd.DataFrame(data)

# Configuración del estilo
plt.style.use('seaborn-v0_8-darkgrid')
plt.figure(figsize=(12, 6))

# Gráfico de barras (promedio por dificultad)
ax = sns.barplot(data=df, x='dificultad', y='nota', 
                 order=['Fácil', 'Media', 'Difícil'], 
                 palette='Blues_d', 
                 ci=None, 
                 estimator=np.mean)


# Personalización
plt.title('Relación entre Dificultad y Nota por Pregunta', fontsize=14, pad=20)
plt.xlabel('Dificultad', fontsize=12)
plt.ylabel('Nota Promedio', fontsize=12)
plt.ylim(0, 5.5)

# Añadir valores exactos
for p in ax.patches:
    ax.annotate(f'{p.get_height():.1f}', 
                (p.get_x() + p.get_width() / 2., p.get_height()), 
                ha='center', va='center', 
                xytext=(0, 10), 
                textcoords='offset points',
                fontsize=10)

# Leyenda personalizada
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], marker='s', color='w', label='Nota promedio',
           markerfacecolor='steelblue', markersize=10)
]
plt.legend(handles=legend_elements, loc='upper right')

plt.tight_layout()
plt.show()
```
